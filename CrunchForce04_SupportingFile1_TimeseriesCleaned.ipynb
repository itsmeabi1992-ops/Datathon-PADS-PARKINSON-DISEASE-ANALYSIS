```python
import pandas as pd
import glob
import json

base = r"C:\Users\vidya\OneDrive\Desktop\DATA ANALYST\NUMPY NINJA\DATATHON\pads-parkinsons-disease-smartwatch-dataset"
```

```python
patients_files = glob.glob(base + r"\patients\*.json")

patients_list = []
for f in patients_files:
    with open(f, 'r') as file:
        patients_list.append(json.load(file))

patients_df = pd.json_normalize(patients_list)
print("Patients shape:", patients_df.shape)
print("Patients columns:", patients_df.columns.tolist())
```

```python
movement_files = glob.glob(base + r"\movement\*.json")

movement_list = []
for f in movement_files:
    with open(f, 'r') as file:
        movement_list.append(json.load(file))

movement_df = pd.json_normalize(movement_list)
print("Movement shape:", movement_df.shape)
print("Movement columns:", movement_df.columns.tolist())
```

```python
questionnaire_files = glob.glob(base + r"\questionnaire\*.json")

questionnaire_list = []
for f in questionnaire_files:
    with open(f, 'r') as file:
        questionnaire_list.append(json.load(file))

questionnaire_df = pd.json_normalize(questionnaire_list)
print("Questionnaire shape:", questionnaire_df.shape)
print("Questionnaire columns:", questionnaire_df.columns.tolist())
```

```python
import json

# Check 1 sample file from questionnaire
with open(questionnaire_files[0], 'r') as f:
    sample_q = json.load(f)
print("=== QUESTIONNAIRE SAMPLE ===")
print(json.dumps(sample_q, indent=2))
```

```python
# Check 1 sample file from movement
with open(movement_files[0], 'r') as f:
    sample_m = json.load(f)
print("=== MOVEMENT SAMPLE ===")
print(json.dumps(sample_m, indent=2))
```

```python
questionnaire_flat = []

for f in questionnaire_files:
    with open(f, 'r') as file:
        data = json.load(file)
    
    # Flatten each question as a separate row
    for item in data['item']:
        questionnaire_flat.append({
            'subject_id'         : data['subject_id'],
            'study_id'           : data['study_id'],
            'questionnaire_name' : data['questionnaire_name'],
            'link_id'            : item['link_id'],
            'question_text'      : item['text'],
            'answer'             : item['answer']
        })

questionnaire_df = pd.DataFrame(questionnaire_flat)
print("Questionnaire shape:", questionnaire_df.shape)
print(questionnaire_df.head())
```

```python
movement_flat = []

for f in movement_files:
    with open(f, 'r') as file:
        data = json.load(file)
    
    # Flatten session → records
    for session in data['session']:
        for record in session['records']:
            movement_flat.append({
                'subject_id'      : data['subject_id'],
                'study_id'        : data['study_id'],
                'device_id'       : data['device_id'],
                'sampling_rate'   : data['sampling_rate'],
                'record_name'     : session['record_name'],
                'rows'            : session['rows'],
                'device_location' : record['device_location'],
                'file_name'       : record['file_name']
            })

movement_df = pd.DataFrame(movement_flat)
print("Movement shape:", movement_df.shape)
print(movement_df.head())
```

```python
# Extract subject_id from source_file column
df['subject_id'] = df['source_file'].str.split('_').str[0]

# Get valid IDs from patients_df
valid_ids = patients_df['id'].unique()

# Orphan check
orphans = df[~df['subject_id'].isin(valid_ids)]
print("Valid Patient IDs :", len(valid_ids))
print("Orphan rows       :", len(orphans))

if len(orphans) > 0:
    print("Orphan subject IDs:", orphans['subject_id'].unique())
else:
    print("No orphans found ✅")
```

```python
df.columns = ['Time', 'Accelerometer_X', 'Accelerometer_Y', 'Accelerometer_Z',
              'Gyroscope_X', 'Gyroscope_Y', 'Gyroscope_Z', 'source_file', 'subject_id']

print("Done:", df.columns.tolist())
```

```python
# source_file: 001_CrossArms_LeftWrist.txt
df['activity']        = df['source_file'].str.split('_').str[1]
df['device_location'] = df['source_file'].str.split('_').str[2].str.replace('.txt', '')

print(df[['source_file', 'subject_id', 'activity', 'device_location']].head(5))
```

```python
df_agg = df.groupby(['subject_id', 'activity', 'device_location']).agg(
    Accel_X_mean = ('Accelerometer_X', 'mean'),
    Accel_Y_mean = ('Accelerometer_Y', 'mean'),
    Accel_Z_mean = ('Accelerometer_Z', 'mean'),
    Gyro_X_mean  = ('Gyroscope_X', 'mean'),
    Gyro_Y_mean  = ('Gyroscope_Y', 'mean'),
    Gyro_Z_mean  = ('Gyroscope_Z', 'mean'),
    row_count    = ('Time', 'count')
).reset_index()

print("Aggregated shape:", df_agg.shape)
print(df_agg.head())
```
